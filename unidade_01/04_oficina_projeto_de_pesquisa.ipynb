## 1. Da motivação à pergunta

Descreva por que o fenômeno importa antes de escrever uma pergunta. Depois,

### Minha formulação

**Motivação humanística:**
Estudos sobre o campo literário mostram que mulheres tendem a estar menos
presentes onde há mais prestígio em jogo. Verboord (2012), por exemplo,
encontrou mulheres mais bem representadas entre best-sellers do que entre
vencedores de prêmios literários na França, na Alemanha e nos Estados Unidos
entre 1960 e 2009. No Brasil, Dalcastagnè (2005) mostrou que 72,7% dos autores
dos 258 romances analisados, publicados por três grandes editoras, eram homens.
Redes sociais de leitores costumam ser apresentadas como espaços mais abertos,
nos quais o reconhecimento viria do público, não de instituições. Vale
perguntar se a atenção registrada nesses espaços também se distribui de forma
desigual.

**Conceito central e autores de referência:**
Reconhecimento literário, entendido aqui como atenção pública registrada.
- Merton (1968) descreve o efeito Mateus: quem já é reconhecido tende a
  receber reconhecimento adicional. Isso ajuda a entender por que contagens de
  avaliações são tão concentradas em poucos livros.
- Rossiter (1993) nomeia o efeito Matilda: o reconhecimento de mulheres na
  ciência é sistematicamente subestimado ou atribuído a homens. Uso a ideia
  como hipótese a investigar no campo literário, não como fato demonstrado.
- Thelwall (2017, 2019) analisou o Goodreads e encontrou, em quase todos os
  gêneros, desequilíbrio na proporção de autoras e autores e, em vários
  gêneros comuns, avaliações mais altas dadas por leitores a autores do próprio
  gênero.
- D'Ignazio e Klein (2020) lembram que números não falam por si: é preciso
  perguntar quem produziu as categorias e o que elas deixam de fora.

**Questão ampla:**
Como desigualdades de gênero atravessam o reconhecimento de obras literárias em
redes sociais de catalogação e avaliação de livros?

**Pergunta delimitada:**
No recorte de 10.454 livros extraídos do Goodreads, livros de autoria
registrada como feminina recebem número de avaliações e de resenhas semelhante
ao de livros de autoria registrada como masculina? A diferença muda conforme o
gênero literário principal?

**Finalidade predominante e justificativa:**
Descritiva. O objetivo é medir e descrever a diferença de reconhecimento no
recorte. Não pretendo explicar suas causas nem prever o sucesso de livros.

**Estrutura analítica inicial, etapas secundárias e justificativa:**
Estrutura comparativa (autoria feminina × masculina), com uma etapa
associativa: examinar se a diferença varia junto com o gênero literário. Etapa
secundária: verificar se a composição dos grupos (ano de publicação, gênero
literário) produz parte da diferença, o que exige comparações dentro de
subconjuntos parecidos.

**Tarefa computacional inicial possível:**
Calcular medianas, quartis e proporções de livros muito avaliados para cada
grupo; repetir dentro de cada gênero literário; comparar a taxa de resenhas por
avaliação; retornar a casos extremos.

**O que mudou em relação ao produto do Notebook 01 e por quê?**
Mantive a pergunta e a finalidade. Acrescentei: (1) a nota média como dimensão
apenas secundária, para não deslizar para julgamento de qualidade; (2) a
verificação da composição dos grupos, porque o resumo do Notebook 03 mostrou
que o recorte mistura obras de −720 com lançamentos de 2019; (3) a expressão
“autoria registrada como”, porque o rótulo de gênero tem origem desconhecida.

como medidas principais. A nota média fica como secundária porque está mais
próxima de um julgamento de qualidade. Para o gênero da autoria, o rótulo do
arquivo será a medida inicial, e a classificação pelo nome servirá para
auditá-lo.

Complete a cadeia lógica:

### Minha cadeia de evidência

**Operação prevista:**
Agrupar os livros pelo gênero registrado da autoria e comparar as medianas de
`n_avaliacoes` e de `n_resenhas`, no conjunto e dentro de cada gênero literário
principal com pelo menos 30 livros em cada grupo.

**Resultado possível:**
A mediana de avaliações dos livros de autoria feminina é menor que a dos livros
de autoria masculina no conjunto, mas a diferença muda de tamanho, ou até de
sentido, conforme o gênero literário.

**De que esse resultado poderia ser evidência?**
De que, neste recorte, a atenção registrada pelos usuários se distribui de forma
desigual entre os grupos, e de que essa desigualdade depende do gênero
literário em que o livro foi classificado.

**O que ele não demonstraria diretamente?**
Que leitores discriminam autoras; que livros de autoras são piores ou melhores;
que o padrão vale para todo o Goodreads; ou a causa da diferença.

**Interpretações alternativas:**
- composição: autoras podem estar concentradas em gêneros que recebem menos
  avaliações em geral (por exemplo, romance);
- tempo de exposição: livros mais antigos, mais frequentes entre homens no
  recorte, acumulam avaliações por mais tempo;
- seleção: o critério desconhecido do recorte pode ter favorecido um grupo;
- classificação: erros no rótulo de gênero podem distorcer os grupos;
- mercado: investimento editorial e marketing desiguais, que não estão nos
  dados.

**Casos ou documentos que precisariam ser relidos:**
Os livros mais avaliados de cada grupo; livros de autoras em gêneros com
predominância masculina (ficção científica, história) e de autores em gêneros
com predominância feminina (romance); casos com rótulo de gênero duvidoso; as
páginas desses livros no Goodreads, incluindo resenhas.

**Conhecimento contextual necessário:**
Funcionamento do Goodreads (estantes, estrelas, resenhas, quem usa a rede);
história do mercado editorial e dos gêneros literários associados a mulheres;
estudos anteriores sobre gênero e recepção (Thelwall, 2017, 2019; Verboord,
2012).

A cadeia descreve uma interpretação possível. Para não convertê-la em

## U01-A13 — Atividade — revisão entre pares

**Modalidade:** duplas, com dois turnos. **Tempo sugerido:** 15 minutos por
proposta. **Produto:** parecer breve recebido e registro das mudanças
realizadas.

Peça a uma colega ou a um colega que responda:

1. consigo identificar exatamente o que será observado?
2. os dados propostos podem responder à pergunta?
3. conceito e indicador estão diferenciados?
4. o alcance da conclusão está claro?
5. qual escolha precisa de maior justificativa?

### Parecer recebido (revisão simulada)

1. **Sim.** Serão observados `n_avaliacoes`, `n_resenhas` e a taxa entre eles,
   por grupo de gênero registrado e por gênero literário.
2. **Em parte.** Os dados respondem à pergunta descritiva sobre o recorte, mas
   não sobre o Goodreads inteiro, porque o critério de seleção é desconhecido.
3. **Sim.** A proposta separa reconhecimento (conceito) de número de avaliações
   e de resenhas (indicadores) e reconhece que nenhum mede qualidade.
4. **Sim**, a expressão “neste recorte” aparece na pergunta e no resumo.
5. **A escolha que mais precisa de justificativa é usar o rótulo
   `genero_autor` sem saber como foi produzido.** Se houver erro sistemático,
   todas as comparações ficam comprometidas. Comentário prioritário: planejar
   uma auditoria do rótulo antes de comparar os grupos.

### Mudanças realizadas e justificativa

Incluí na “Próxima decisão necessária” a auditoria do rótulo de gênero como
condição para as comparações. Mantive a pergunta, porque o parecer não apontou
problema de delimitação. Acrescentei ao esquema o campo `genero_por_nome`, que
