## 1. Alinhamento

**Pergunta delimitada:** no recorte de 10.454 livros extraídos do Goodreads,
livros de autoria registrada como feminina recebem número de avaliações e de
resenhas semelhante ao de livros de autoria registrada como masculina? A
diferença muda conforme o gênero literário principal?

**Finalidade e estrutura analítica:** finalidade descritiva; estrutura
comparativa (dois grupos de livros) com etapa associativa (variação da
diferença conforme o gênero literário).

**Unidade de análise:** o livro (`id_livro`).

**Que observações seriam necessárias para responder:** para cada livro, o
número de avaliações e de resenhas, o gênero literário principal, o ano de
publicação e um gênero da autoria confiável. Para interpretar, seriam úteis a
data da coleta e o critério de seleção dos livros, que estão ausentes.

Com pergunta e unidade de análise definidas, identifique quais fontes podem
fornecer essas observações e reconstrua a cadeia que as produziu.

## 2. Fontes e cadeia de produção

**Fontes primárias em relação à pergunta:** os registros de uso do Goodreads
(contagens de avaliações e resenhas, notas, estantes que geram os gêneros
literários), acessados por meio do CSV recebido.

**Fontes secundárias e dados derivados:** o CSV é um dado derivado (extraído,
selecionado e traduzido por terceiros). Fontes secundárias: Thelwall (2017,
2019), Verboord (2012), Dalcastagnè (2005), D'Ignazio e Klein (2020).

**Instituições, produtores e custodiantes:** usuários do Goodreads produzem
avaliações, resenhas e estantes; a empresa mantém a rede (pertencente à
Amazon desde 2013, segundo Slashdot, 2020); um extrator não identificado
produziu o recorte; o estudante custodia a cópia local.

**Finalidade original dos registros e categorias herdadas:** as avaliações
servem para que usuários organizem leituras e recomendem livros, e a rede usa
esses dados para recomendação e descoberta. Categorias herdadas: gêneros
literários das estantes dos usuários; rótulo binário de gênero da autoria, de
origem desconhecida; local de nascimento em texto livre.

**Condições materiais e técnicas de acesso:** a cópia local está disponível. Uma
nova coleta é difícil: a API deixou de emitir novas chaves em dezembro de 2020
(Slashdot, 2020), e a raspagem de páginas depende dos termos de uso, que
precisam ser lidos antes de qualquer coleta.

Conhecer as fontes não basta para formar a base. Delimite agora a população
de interesse, o universo acessível e o corpus, justificando cada regra de
inclusão e exclusão.

## 3. População, corpus e seleção

**População de interesse:** livros catalogados no Goodreads e o reconhecimento
que recebem dos usuários.

**População acessível:** os 10.454 livros do arquivo recebido.

**Corpus previsto:** os mesmos 10.454 livros na análise principal; subconjuntos
registrados para as análises de sensibilidade.

| Critério | Campo/evidência | Regra | Justificativa | Caso limítrofe |
|---|---|---|---|---|
| inclusão | `id_livro` | todos os livros do arquivo, sem filtro | critério original desconhecido; filtrar antes de entender esconderia decisões | títulos repetidos de autores diferentes (18 livros): mantidos, são obras distintas |
| exclusão (sensibilidade) | `ano_publicacao` | retirar livros anteriores a 1990 | reduzir diferença de tempo de exposição entre grupos | clássicos com ano negativo (ex.: −720): saem nesta análise, ficam na principal |
| exclusão (sensibilidade) | `genero_autor` auditado | retirar livros com rótulo duvidoso (definido na Unidade 3) | evitar comparar grupos contaminados por erro de classificação | pseudônimos e nomes com iniciais: não são duvidosos só por não terem prenome classificável |
| exclusão (comparações por gênero literário) | `genero_1` | gêneros com menos de 30 livros em algum grupo ficam fora dessa comparação | medianas com poucos casos são instáveis | gêneros com 30 a 50 livros: incluídos, com cautela |

**Registro das exclusões:** tabela de motivos com uma coluna booleana por
critério e uma coluna de inclusão por análise, gerada pelo código e salva com a
base derivada.

Toda seleção altera presenças e ausências. Use os critérios recém-definidos
para avaliar cobertura, possíveis vieses e o alcance legítimo das
afirmações.

## 4. Cobertura, vieses e silêncios

| Dimensão | Cobertura | Lacuna | Efeito sobre a análise | Mitigação |
|---|---|---|---|---|
| temporal | −720 a 2019; 52% entre 2010 e 2019 | nada após 2019; livros antigos mais frequentes entre autores homens | idade do livro pode imitar efeito de gênero | sensibilidade com 1990–2019 |
| espacial | cerca de 62% EUA e 12% Reino Unido (valores brutos) | literaturas não anglófonas quase ausentes | resultados valem para um universo anglófono | declarar; comparar nascidos e não nascidos nos EUA |
| social | autoria feminina e masculina | pessoas não binárias; leitores invisíveis | comparação binária; nada sobre quem reconhece | declarar o apagamento |
| institucional | registros do Goodreads | proveniência e seleção desconhecidas | possível viés de seleção não mensurável | pedir proveniência; não generalizar |
| documental | contagens e notas | resenhas, livros com menos de 5 avaliações | só livros já visíveis; tipo de reconhecimento desconhecido | interpretar como distribuição entre livros visíveis |

**Possíveis mecanismos de viés de seleção:** se o recorte foi montado a partir de
listas de livros populares ou de autores muito avaliados, ele exclui a cauda de
livros pouco lidos, onde a desigualdade pode ser diferente. O limite de seis
livros por autor reduz o peso de autores prolíficos. A coleta anglófona
favorece o mercado editorial dos EUA. Nenhum desses mecanismos pode ser
confirmado sem a proveniência.

**Silêncios não redutíveis a valores ausentes:** pessoas não binárias, que o
campo não permite registrar; livros que nunca chegaram ao Goodreads; obras de
autoras que não foram publicadas ou traduzidas; o conteúdo do reconhecimento
(o que as resenhas dizem).

**Limites das afirmações:** só afirmo diferenças descritivas entre grupos
neste recorte, com o rótulo de gênero auditado, e sempre acompanhadas das
análises por período e por gênero literário.

Esses limites precisam permanecer ligados aos registros. Defina os
metadados, identificadores e informações de proveniência que tornarão as
decisões auditáveis.

## 5. Metadados, identificadores e proveniência

**Identificador, escopo e persistência:** `id_livro` (único no recorte) e
`id_autor` (único na tabela de autores), ambos herdados do Goodreads.
Persistência dependente do Goodreads; por isso, guardo também título e autor.

| Campo | Definição | Tipo/domínio | Origem | Regra | Limitação |
|---|---|---|---|---|---|
| `n_avaliacoes` | avaliações por estrelas do livro | inteiro ≥ 5 | uso da rede | usar como recebido | data de coleta desconhecida |
| `n_resenhas` | resenhas escritas | inteiro ≥ 0, ≤ `n_avaliacoes` | uso da rede | usar como recebido | sem conteúdo |
| `taxa_resenhas` | resenhas por avaliação | proporção de 0 a 1 | derivado | `n_resenhas / n_avaliacoes` | instável com poucas avaliações |
| `genero_autor` | gênero registrado da autoria | feminino, masculino | herdado | manter original | regra desconhecida; binário |
| `genero_por_nome` | classificação do prenome | categorias do dicionário de nomes | derivado | primeiro prenome sem iniciais | não identifica pessoas; erra pseudônimos |
| `pais_padronizado` | país de nascimento padronizado | texto ou ausente | derivado de `pais_nascimento` | tabela de equivalências explícita | escalas e entidades históricas |

O dicionário completo está em `projeto_goodreads/dados/dicionario_dados.csv`.

**Metadados administrativos e de direitos:** responsável pela cópia, data de
recebimento, hash, termos de uso a verificar, ausência de licença declarada.

**Como a fonte será ligada ao registro derivado:** pelos ids do Goodreads e
pelo par título–autor; colunas originais preservadas ao lado das derivadas.

**Versão, data de acesso, agente e transformações:** versão identificada pelo
hash SHA-256; recebida em 15/10/2026; agente: o estudante; transformações
registradas em `preparar_base.py` e no log de execução.

A documentação técnica permite reconstruir a base, mas não resolve por si
só quem pode acessá-la, reutilizá-la ou ser afetado por ela. Examine agora

## 6. Governança, reuso, ética e questões legais

Acesso público não significa ausência de responsabilidade. Identifique,
sem emitir parecer jurídico:

**Evidências e lacunas FAIR prioritárias:** evidências: ids únicos, dicionário,
hash e proveniência parcial. Lacunas prioritárias: protocolo de acesso, licença
e regra de atribuição do gênero.

**Forma de acesso, licença ou termos de reutilização:** cópia local de uso
restrito ao projeto; licença desconhecida; termos de uso do Goodreads a
verificar antes de qualquer redistribuição.

**Usos recomendados e usos que devem ser evitados:** recomendados: análise
agregada, ensino, formulação de hipóteses. Evitar: rotular pessoas, afirmar
qualidade literária, generalizar, uso comercial.

**Responsável pelo datasheet, manutenção e novas versões:** o estudante.

**Dados pessoais ou pessoais sensíveis possíveis:** nomes de autores são dados
pessoais (LGPD, art. 5º, I). O gênero não aparece na lista de dados sensíveis
do art. 5º, II, que inclui, entre outros, origem racial ou étnica, convicção
religiosa, opinião política e dado referente à saúde ou à vida sexual. Ainda
assim, atribuir gênero a pessoas sem consentimento pede cautela, e alguns
autores podem ter falecido ou usar pseudônimo.

**Pessoas ou comunidades potencialmente afetadas:** autores e autoras do
recorte; usuários que produziram avaliações; leitores dos resultados.

**CARE é pertinente ao caso? Se houver dados indígenas, como benefício,
autoridade, responsabilidade e ética serão definidos com os povos
envolvidos?** Não declaro pertinência: o projeto não trata de dados,
conhecimentos ou territórios indígenas, e não há campo que identifique autoria
indígena. Se a análise passar a destacar obras de autoria indígena, a decisão
precisará ser revista com participação das pessoas e comunidades envolvidas.

**Necessidade de avaliação pelo sistema de ética da instituição:** a análise
usa dados agregados de acesso público, sem contato com pessoas. Se houver
publicação ou coleta de resenhas, consultar o comitê de ética da instituição
para saber se o projeto precisa de avaliação.

**Finalidade, minimização, segurança e controle de acesso:** finalidade
acadêmica declarada; uso apenas dos campos necessários; cópia local sem
publicação; a auditoria de gênero por nome fica restrita ao projeto.

**Direitos autorais, licença, termos de uso e autorização:** o arquivo não traz
texto de resenhas nem de livros, o que reduz questões de direito autoral.
Termos de uso do Goodreads e da fonte intermediária ainda não verificados.

**Forma de publicação e risco de reidentificação:** publicar apenas tabelas
agregadas por grupo e gênero literário; casos individuais apenas de livros e
autores públicos, quando indispensáveis. Os autores já são identificados pelo
nome; o risco relevante não é reidentificação, mas exposição a rótulos
indevidos.

Para situações concretas, consulte a instituição, o comitê de ética, a
ANPD e assessoria jurídica quando aplicável. O guia da ANPD não possui
caráter normativo e não substitui a legislação.

As responsabilidades identificadas precisam caber nas condições reais do
projeto. Confronte o desenho desejado com tempo, acesso, infraestrutura e
alternativas de contingência.

## 7. Viabilidade e contingência

**Estimativa de volume e trabalho de coleta:** nenhuma coleta nova; 10.454
linhas, 1,3 MB; processamento em segundos em um computador comum.

**Amostra piloto:** os autores mais avaliados de cada tipo de divergência
entre rótulo e nome (12 de cada lado, na execução da Unidade 3), para conferir
manualmente o rótulo de gênero.

**Dependências de autorização ou infraestrutura:** nenhuma para a análise
local; verificação dos termos de uso para publicar; pacote `gender-guesser`
para a auditoria por nome.

**Plano B se a principal fonte não puder ser usada:** usar bases acadêmicas
documentadas sobre o Goodreads, como o conjunto descrito por Sabri e Weber
(2021), com proveniência publicada, aceitando mudar o recorte.

**O que será deliberadamente deixado de fora:** qualidade literária; texto das
resenhas; perfil dos leitores; causas da desigualdade; outras redes e o
mercado editorial.

antes de submetê-lo à leitura de outra pessoa.

## U02-A11 — 9. Revisão por pares

O colega deve localizar no protocolo:

1. uma escolha bem fundamentada;
2. uma exclusão que possa alterar as conclusões;
3. um campo ou metadado ainda ambíguo;
4. um risco ético ou legal que precise de verificação;
5. uma afirmação que exceda a cobertura.

**Parecer recebido (revisão simulada):**
1. *Escolha bem fundamentada:* manter todos os livros na base principal e
   tratar exclusões como sensibilidade registrada.
2. *Exclusão que pode alterar conclusões:* retirar livros com rótulo
   duvidoso. Se os erros se concentrarem em um grupo, a exclusão muda o
   tamanho e a composição desse grupo.
3. *Campo ambíguo:* `ano_publicacao` não diz se é a primeira publicação ou a
   edição catalogada.
4. *Risco a verificar:* os termos de uso do Goodreads para dados extraídos por
   terceiros.
5. *Afirmação que excede a cobertura:* a motivação fala em “redes sociais de
   leitores”, no plural, mas a base cobre uma só rede.

**Mudanças realizadas e justificativa:** (1) a exclusão de rótulos duvidosos
será acompanhada de duas alternativas (rótulo original e reclassificação),
para mostrar se a conclusão depende dela; (2) registrei no dicionário a dúvida
sobre `ano_publicacao`; (3) mantive o risco dos termos de uso como pendência;
(4) restringi a pergunta e as conclusões ao Goodreads, deixando o plural apenas
na questão ampla.
