## Como realizar a oficina

![Pergunta, tabela, figura, leitura e limite formam uma cadeia; divergências exigem revisão.](imagens/04_cadeia_argumento.svg)

Siga esta ordem:

1. declare a pergunta, o corpus e a unidade de análise;
2. selecione somente medidas pertinentes às variáveis e à pergunta;
3. produza tabelas antes de interpretar gráficos;
4. descreva o padrão observado sem explicar sua causa;
5. retorne a documentos ou trechos que possam qualificar o agregado;
6. formule hipóteses alternativas e indique dados necessários para avaliá-las;

Para cada resultado, use a sequência abaixo:

- **Procedimento:** o que foi calculado, sobre quais registros e com qual regra?
- **Descrição:** que valores ou padrões aparecem?
- **Interpretação:** que leitura situada pode ser considerada?
- **Limite:** o que esse resultado não permite afirmar?
- **Próximo passo:** que caso, dado ou método deveria ser examinado?

Evite frases como “o gráfico prova”. Prefira “nesta base”, “o padrão observado” e
“uma hipótese a investigar”.

A sequência anterior funciona como um protocolo geral. Sua aplicação começa
pela delimitação: antes de calcular ou visualizar, é necessário saber quais casos
estão incluídos e o que cada linha permite observar.

**Minha análise quantitativa:** neste recorte, os livros de autoria feminina
têm menor alcance no centro da distribuição: mediana de 3.129 avaliações contra
5.218 dos livros de autoria masculina (E2). O número de resenhas é parecido
(medianas de 320 e 361), o que significa que a **taxa de resenhas por avaliação
é maior entre livros de autoras** (0,114 contra 0,076). Notas médias são
praticamente iguais (3,92 e 3,97). A diferença de alcance existe nas três
especificações (razão entre medianas de 0,71 em E1, 0,60 em E2 e 0,56 em E3) e
diminui quando se consideram só livros publicados de 1990 a 2019 (0,84; 0,72;
0,70). A composição por gênero literário é muito desigual. Hipótese provisória:
idade dos livros e gênero literário explicam parte da diferença de alcance.

O perfil quantitativo descreve composição, extensão e cruzamentos, mas não
mostra como os temas são formulados nos textos. A exploração textual acrescenta
essa camada sem abandonar os IDs, denominadores e limites já estabelecidos.

## 3. Perfil textual

| Etapa | Regra ou parâmetro | Resultado selecionado | Interpretação e limite |

**Meu perfil textual:** como o único texto do arquivo é o título, o perfil
textual mostra como os livros são nomeados. Títulos de autoras usam mais
*love*, *girl*, *woman*, *heart* e *secret*; títulos de autores, mais *war*,
*man*, *world* e *history*. A lista de palavras funcionais retirou pronomes
possessivos como *her* e *his*, que poderiam indicar marcas de gênero; essa
escolha reduz o que o perfil consegue mostrar. A tokenização partiu contrações
(“don t”), o que aparece nas colocações. A leitura das concordâncias mostra que
essas palavras seguem convenções de gênero literário e de mercado (romance,
jovem adulto, história militar), e não um “estilo feminino” ou “masculino”.

Os perfis quantitativo e textual fornecem vários resultados possíveis. A etapa
seguinte não consiste em desenhar todos eles, mas em selecionar aqueles que melhor
respondem à pergunta e comunicá-los sem romper a ligação com as tabelas.

Uma hipótese exploratória só é intelectualmente útil quando suas fronteiras
podem ser examinadas. A próxima seção registra tanto os limites da evidência
quanto o percurso necessário para que outra pessoa reconstrua os resultados.

## 7. Limitações e reprodutibilidade

### Limitações

**Limitações do relatório:**
- *Sub-representação:* literaturas não anglófonas, livros com menos de cinco
  avaliações, livros posteriores a 2019, pessoas não binárias.
- *Escolhas que afetam os resultados:* a especificação de gênero (E1, E2, E3)
  muda o tamanho da diferença; a lista de palavras funcionais retirou *her* e
  *his*; a tokenização partiu contrações; as faixas de ano e de avaliações foram
  escolhidas por mim; a regra de 1,5 IQR só funciona em escala log.
- *Padrões possivelmente produzidos pela construção da base:* o limite de seis
  livros por autor, o mínimo de cinco avaliações e a concentração em 2010–2014
  podem vir do critério de seleção desconhecido; o erro do rótulo de gênero foi
  produzido por quem montou o arquivo.
- *Conclusões que não devem ser formuladas:* que o Goodreads discrimina
  autoras; que livros de autoras são piores ou melhores; que a diferença vale
  fora do recorte; que o gênero inferido pelo nome é o gênero das pessoas.
