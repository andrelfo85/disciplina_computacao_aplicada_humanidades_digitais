## 1. Pergunta e contexto

**Pergunta comparativa:** neste recorte do Goodreads, livros de autoria
registrada como feminina recebem o mesmo alcance (número de avaliações) que
livros de autoria registrada como masculina? A diferença se mantém quando se
comparam livros do mesmo período e do mesmo gênero literário?

**Contexto histórico, social, linguístico ou cultural:** o Goodreads é uma rede
social de catalogação e avaliação de livros em que usuários registram leituras,
atribuem estrelas e escrevem resenhas. O recorte é dominado por livros em
inglês e por autores nascidos nos EUA e no Reino Unido, com 39% dos livros
publicados entre 2010 e 2014 e um conjunto de clássicos antigos.

**Literatura que torna a comparação pertinente:** Verboord (2012) mostra que a
desigualdade de gênero é maior onde há mais prestígio em jogo; Thelwall (2017,
2019) documenta desequilíbrio de autoria entre gêneros literários no Goodreads e
diferenças de avaliação ligadas ao gênero de leitores e autores; Merton (1968)
e Rossiter (1993) descrevem a vantagem cumulativa no reconhecimento e sua
desigualdade de gênero; Underwood, Bamman e Lee (2018), em um estudo com mais de
100 mil romances em inglês, relatam que o espaço ocupado por mulheres na ficção
caiu pela metade entre o século XIX e 1960 (segundo a divulgação do estudo pela
Universidade de Illinois; convém conferir no artigo).

**Por que comparar estes grupos ou documentos:** porque a pergunta da pesquisa
é sobre a distribuição do reconhecimento conforme o gênero da autoria. A
comparação por período e por gênero literário é necessária porque a Unidade 4
mostrou que os grupos diferem muito em idade dos livros e em gênero
literário.

A pergunta delimita o contraste; agora verifique se as unidades e os
subconjuntos são comparáveis e quais assimetrias precisam permanecer visíveis.

## 3. Especificação principal

**Percurso escolhido:** quantitativo (com complemento textual sobre os títulos).

**Quantidade, representação ou variável:** `log10_avaliacoes`, logaritmo de
base 10 do número de avaliações de cada livro.

**Medida principal:** diferença de médias de log10 das avaliações, F − M,
calculada (a) no conjunto e (b) dentro de cada ano de publicação de 1990 a
2018, resumida pela média ponderada pelo número de livros de cada ano.

**Fórmula ou convenção implementada:**
Δ = média(log10 avaliações | F) − média(log10 avaliações | M);
Δ_ano = Σ_t w_t Δ_t / Σ_t w_t, com w_t = número de livros do ano t (anos com
pelo menos 10 livros em cada grupo). A razão de médias geométricas é 10^Δ.

**Pré-processamento e parâmetros:** especificação E2; livros com rótulo duvidoso
excluídos; mínimo de 10 livros por grupo em cada ano e de 30 em cada gênero
literário; bootstrap com B = 2.000, reamostrando autores; permutação com
M = 5.000; semente 20261015.

**Por que esta medida responde à pergunta:** o logaritmo trata a assimetria das
contagens e permite ler a diferença como razão (10^Δ). A comparação dentro de
cada ano responde à segunda parte da pergunta: se a diferença persiste entre
livros que tiveram tempo de exposição parecido.

## 5. Especificação alternativa e sensibilidade

**Alternativa escolhida:** mediana (no lugar da média em log) e especificações
de gênero E1 e E3 (no lugar de E2).

**Razão metodológica:** a mediana não depende de transformação; as
especificações de gênero testam se a decisão sobre os rótulos duvidosos muda a
conclusão.

**Resultado alternativo:**

| Especificação | Razão de medianas, todos os anos | Razão de medianas, 2010–2019 |
|---|---:|---:|
| E1 (rótulo original) | 0,71 | 1,34 |
| E2 (principal) | 0,60 | 1,46 |
| E3 (duvidosos como feminino) | 0,56 | 1,41 |

**O que permaneceu e o que mudou:** permaneceram o sentido da diferença
agregada (negativa para livros de autoras) e a inversão entre livros recentes
(positiva para livros de autoras). Mudou a magnitude: a auditoria do rótulo
aumenta a diferença agregada (de 0,71 para 0,60), porque o rótulo original
colocava livros populares de autoras no grupo masculino.

**A conclusão depende da escolha?** A *direção* não depende nem da medida nem da
especificação de gênero. Depende, sim, do **recorte temporal**: a afirmação
“livros de autoras recebem menos avaliações” só vale para o conjunto, que
mistura clássicos antigos e lançamentos recentes. A taxa de resenhas é o caso
mais sensível: a vantagem agregada das autoras (+0,038) desaparece quando se
compara dentro de cada ano (−0,009).

A sensibilidade compara agregados. Retorne agora aos registros ou trechos que
sustentam, contradizem ou explicam a divergência entre medidas.

## 7. Interpretação, relevância e limites

**Afirmação diretamente sustentada:** neste recorte de 9.180 livros do
Goodreads (especificação E2), livros de autoria registrada como feminina têm
mediana de 3.129 avaliações e livros de autoria masculina, 5.218; entre livros
publicados de 2010 a 2019, a relação se inverte (razão de 1,46 entre as
medianas).

**Relevância substantiva da diferença ou semelhança:** a diferença agregada é
grande na escala original (cerca de 2.100 avaliações na mediana), mas não indica
que, entre livros comparáveis em idade, os de autoras recebam menos atenção. A
relevância está justamente na decomposição: a desigualdade observada se
concentra no cânone acumulado e em gêneros como ficção científica e não ficção.

**Interpretação contextual:** a leitura mais compatível com os dados é que a
rede reflete uma **desigualdade histórica acumulada**: clássicos antigos,
majoritariamente de autores, reúnem milhões de avaliações por estarem em
currículos, coleções e listas. Isso dialoga com a vantagem cumulativa de Merton
(1968). Entre lançamentos recentes, livros de autoras recebem mais avaliações em
quase todos os gêneros, o que é coerente com a forte presença feminina na
leitura e na produção de gêneros como romance e jovem adulto. Essas leituras são
hipóteses: o arquivo não informa quem avalia nem como os livros foram
selecionados.

**Hipóteses alternativas:** (1) o critério de seleção do recorte pode ter
incluído livros recentes de autoras e de autores por regras diferentes; (2)
livros recentes de autores podem estar mais presentes em edições ou
identificadores diferentes, dividindo avaliações; (3) parte da diferença
recente pode refletir campanhas editoriais e comunidades de leitores de gêneros
específicos, e não o gênero da autoria.

**O que não pode ser concluído:** que o Goodreads favorece ou desfavorece
autoras; que livros de um grupo são melhores; que a diferença tem causa no
gênero da autoria; que o padrão vale para outras redes, para o mercado
editorial ou para a crítica.

**Limites de seleção, medição, representação e generalização:** seleção
desconhecida; rótulo de gênero binário e com erros, auditado por um dicionário
de nomes que também erra; “alcance” medido apenas por contagens; resultados
restritos a um recorte anglófono e até 2019.

Antes da leitura externa, avalie se o argumento mantém coerência entre pergunta,
medida, casos e conclusão.

## 10. Síntese e entrega

**Resumo da pergunta, método e resultado:** perguntei se livros de autoria
registrada como feminina recebem, em um recorte de 10.454 livros do Goodreads,
o mesmo alcance que livros de autoria masculina. Depois de auditar o rótulo de
gênero (que classificava como “masculino” centenas de autoras), comparei a
média de log10 das avaliações entre 4.638 livros de autoras e 4.542 de autores.
No conjunto, livros de autoras têm cerca de dois terços das avaliações dos de
autores (Δ = −0,176; intervalo [−0,225; −0,124]), resultado estável sob
reamostragem, o que não indica, por si, importância.

**Como a especificação alternativa alterou ou confirmou a leitura:** a mediana
e as especificações E1 e E3 confirmaram a direção e mudaram a magnitude. A
comparação dentro de cada ano mudou a leitura: entre livros do mesmo ano, os de
autoras recebem mais avaliações (+0,131 em log10; +0,192 em 2010–2019).

**Como os casos qualificaram o agregado:** o livro mais avaliado do recorte é de
uma autora; o cânone antigo é majoritariamente masculino; e a ficção científica
mantém parte da desvantagem mesmo entre livros recentes.

**Conclusão substantiva limitada:** neste recorte, a desigualdade de gênero no
reconhecimento aparece sobretudo como herança de um cânone acumulado e em
gêneros de predominância masculina, e não como desvantagem generalizada de
livros recentes de autoras. A taxa de resenhas por avaliação não difere de
forma consistente quando se compara dentro de cada ano. As conclusões valem só
para o recorte, cuja seleção é desconhecida.

**Mudanças realizadas após a revisão:** limites ampliados (assimetria da E2),
frase sobre estabilidade corrigida, hipótese sobre leitoras marcada como
hipótese, comparação por autor registrada como próximo passo.
